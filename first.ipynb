# Type Conversion

```python
# Implicite vs Explicite type conversion
print(5+5.6)

print(type(5),type(5.6))
# Python automatically converts the integer 5 to a float 5.0 before performing the addition, this is called implicit type conversion.

print(4 + "4") # This will give an error because we cannot add an integer and a string without explicit type conversion
```

```python
# Explicit
# Str -> Int
int("4") #  This will convert the string "4" to an integer 4
```

```python
# explicit more eg logical

# This will convert the complex number 4+5j to an integer 4, it will ignore the imaginary part , which means it will take the real part and ignore the rest .

int(4+5j)
 
```

```python
# Explicit
# int to str
str(5) # This will convert the integer 5 to a string "5"

print(type(str(5)))
```
